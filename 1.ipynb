```python
import pandas as pd

# 例: stats_bat.csv 読み込み
df = pd.read_csv('stats_bat.csv', encoding='cp932')

# 各チーム盗塁合計
sb_team = df.groupby('team_name', as_index=False)['sb'].sum()

# 別途用意した勝率データ（例：2021年セ・リーグ）
summary = pd.DataFrame({
    'team_name': ['巨人','阪神','中日','DeNA','広島','ヤクルト'],
    'win_rate': [0.549, 0.589, 0.500, 0.460, 0.488, 0.615]
})

# 合体
result = pd.merge(sb_team, summary, on='team_name')
print(result)
```

```python
print('stats_bat.csvのチーム名:', df['team_name'].unique())
print('summaryのチーム名:', summary['team_name'].unique())
```

```python
import sqlite3

# DBファイル（なければ作成されます）
conn = sqlite3.connect('teams.db')

# pandasの .to_sql を使う
# 'results' というテーブル名で保存（なければ自動作成、あれば上書き）
result.to_sql('results', conn, if_exists='replace', index=False)

conn.close()
```
