```python
import pandas as pd

df = pd.read_csv('stats_pit.csv', encoding='cp932')  # または encoding='shift_jis'
print(df.head())
print(df.columns.tolist())
```

```python
import pandas as pd


# チームごとに合計（hのみ）
team_h = df.groupby('team_name')['h'].sum().reset_index()
print('-- 被安打（h）チームごと --')
print(team_h)

# チームごとに合計（hrのみ）
team_hr = df.groupby('team_name')['hr'].sum().reset_index()
print('-- 被本塁打（hr）チームごと --')
print(team_hr)
```
